- Correctly handle both uppercase and lowercase vowels

# 1. Expected vs Actual Behavior


### **Expected Behavior**

Examples:
- "Hello"  2

### **Actual Behavior**
The original function only checks lowercase vowels:

vowels = "aeiou"

Because of this, uppercase vowels are ignored.

Example:  
"Audio" → returns 3 instead of 4  

# 2.	Add prints to trace execution

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    print("Input text:", text)

    for letter in text:
        print("Checking letter:", letter) 
        
#this only checks the input as it is without changing case sensitivity ( that is why the program doesnot read UPPERCASE letter)
        
        if letter in vowels:
            count += 1
            print("Vowel found:", letter)
            print("Current count:", count)

    print("Final vowel count:", count)
    return count


# User input
text = input("Enter a word or sentence: ")

print("Number of vowels:", count_vowels(text))
```

**d. print("Vowel found:", letter)** = Shows which letters are counted as vowels.

# 3. Hypothesis About the Cause
Answer:

**Hypothesis 1: The function only checks lowercase vowels**


The function uses:    vowels = "aeiou"  This means uppercase vowels like **A, E, I, O, U** are not included in the comparison.  
If the input contains uppercase letters like **"Audio"**, the letter **"A" may be ignored**, causing incorrect results.



**Hypothesis 2: Input assumptions do not match implementation**

The function may have been written assuming all input would be lowercase.However, users may enter mixed-case text such as **"Audio"**, where uppercase letters exist. This causes the function to behave incorrectly during testing.

# 4. Root Cause / Problem

Answer:
The main root cause is that the function does not handle case sensitivity, causing uppercase vowels to be ignored during comparison.  
i.e. case-sensitivity.

Since vowels are defined as:
vowels = "aeiou"  {uppercase letters like "A", "E", "I", "O", "U" are not counted.}

→ So, In the word "Audio" it contain uppercase letter "A" , which is not counted even if it is vowel.
→ The function does not convert input letters to lowercase before checking  

## **5. Implement Simple Fix**

### **Fix Explanation**

- **Fix 1:** Added `.lower()` to each character in the loop  
  → `letter.lower()` is used instead of `letter`  
  → This ensures uppercase letters are converted to lowercase before comparison  

- **Fix 2:** Updated condition check  
  → Changed from `if letter in vowels:`  
  → To `if letter.lower() in vowels:`  
  → This allows both uppercase and lowercase vowels to be detected  

- **Fix 3:** No changes needed to vowel list  
  → `vowels = "aeiou"` remains the same

  


```python

## **Corrected Code**


def count_vowels(text):
    vowels = "aeiou"
    count = 0

    # Input debug
    print("Input text:", text)

    for letter in text:
        # Step tracking
        print("Checking letter:", letter)

        # Case-insensitive check (root fix)
        if letter.lower() in vowels:
            count += 1
            print("Vowel found:", letter)
            print("Current count:", count)

    # Final output
    print("Final vowel count:", count)

    return count


# User input
text = input("Enter a word or sentence: ")

print("Number of vowels:", count_vowels(text))
```

## **4. Test Cases**

## **5. Assertions**


assert count_vowels("Hello") == 2  
assert count_vowels("HELLO") == 2  
assert count_vowels("Education") == 5  
assert count_vowels("AEIOU") == 5  
assert count_vowels("") == 0

**These tests verify:**  
normal inputs,   
uppercase inputs,  
mixed-case inputs,   
edge cases,    
and empty input handling.

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for letter in text:
        # Convert each letter to lowercase before checking
        if letter.lower() in vowels:
            count += 1

    return count


# TEST CASES

print('"Hello" =', count_vowels("Hello"))
print('"HELLO" =', count_vowels("HELLO"))
print('"Education" =', count_vowels("Education"))
print('"AEIOU" =', count_vowels("AEIOU"))
print('"Python" =', count_vowels("Python"))
print('"" =', count_vowels(""))



# ASSERTIONS


assert count_vowels("Hello") == 2
assert count_vowels("HELLO") == 2
assert count_vowels("Education") == 5
assert count_vowels("AEIOU") == 5
assert count_vowels("") == 0

print("\nAll test cases passed successfully.")
```
