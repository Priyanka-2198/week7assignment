# 5. Executable Verification Tests (Assertions)
AI assisted with implementation details, while the test strategy and expected behaviour were defined beforehand.


**My AI Prompt :**

" Create a beginner-friendly Python function called format_phone_number(number).

The function should:
- Accept phone numbers in different formats
- Allow spaces, dashes, and brackets
- Check that the phone number contains exactly 10 digits
- Return the phone number in this format:
(XXX) XXX-XXXX
- Return "Error: Invalid phone number" for invalid input

Use simple Python code with comments.

Also include assertion tests for:
- Valid phone numbers
- Invalid phone numbers
- Edge cases like extra spaces"

**AI - Generated Code:**

````python
def format_phone_number(number):
    import re

    # Reject alphabetic characters
    if re.search(r'[A-Za-z]', number):
        return "Error: Invalid phone number"

    # Extract digits only
    digits = re.sub(r'\D', '', number)

    # Validate length
    if len(digits) != 10:
        return "Error: Invalid phone number"

    # Return standardized format
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"


# Happy path tests
assert format_phone_number("1234567890") == "(123) 456-7890"
assert format_phone_number("123-456-7890") == "(123) 456-7890"
assert format_phone_number("(123) 456 7890") == "(123) 456-7890"

# Edge case test
assert format_phone_number(" 1234567890 ") == "(123) 456-7890"

# Invalid input tests
assert format_phone_number("12345") == "Error: Invalid phone number"
assert format_phone_number("123456789012") == "Error: Invalid phone number"
assert format_phone_number("abc-def-ghij") == "Error: Invalid phone number"
assert format_phone_number("") == "Error: Invalid phone number"


# User input
phone = input("Enter a phone number: ")

# Display result
print("Formatted Phone Number:", format_phone_number(phone))
````


# 6. Simplification and Refinement

As the test architect, I decided to refine the implementation by asking AI to simplify the code while keeping the same behaviour and test coverage.

Even though the original implementation using regular expressions (`re.sub`) worked correctly, I evaluated that it was more advanced than necessary for beginner-level learning.

The simplified version:

- Uses basic string functions instead of regular expressions  
- Is easier to read and debug  
- Is more beginner-friendly  
- Still passes all required tests

# Simplified Beginner-Friendly Implementation


### My AI prompt : 
Can you simplify the implementation and make it more beginner-friendly while keeping the same functionality and test cases?

### AI reply code:

````python
def format_phone_number(number):
    import re

    # Reject alphabetic characters
    if re.search(r'[A-Za-z]', number):
        return "Error: Invalid phone number"

    # Extract digits only
    digits = re.sub(r'\D', '', number)

    # Validate length
    if len(digits) != 10:
        return "Error: Invalid phone number"

    # Return standardized format
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"


# Happy path tests
assert format_phone_number("1234567890") == "(123) 456-7890"
assert format_phone_number("123-456-7890") == "(123) 456-7890"
assert format_phone_number("(123) 456 7890") == "(123) 456-7890"

# Edge case test
assert format_phone_number(" 1234567890 ") == "(123) 456-7890"

# Invalid input tests
assert format_phone_number("12345") == "Error: Invalid phone number"
assert format_phone_number("123456789012") == "Error: Invalid phone number"
assert format_phone_number("abc-def-ghij") == "Error: Invalid phone number"
assert format_phone_number("") == "Error: Invalid phone number"


# User input
phone = input("Enter a phone number: ")

# Display result
print("Formatted Phone Number:", format_phone_number(phone))
````


**This version is simpler because:**
